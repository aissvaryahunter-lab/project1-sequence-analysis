## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
  """ check if the sequence validate the reference and same length"""
    # Write your code here.
  seq_normal = sequence.upper()
  errors = []
  valid_bases = {"A", "C", "G", "T"}
  if len(seq_normal) != len(reference):
        errors.append("wrong_length")
  if any(base not in valid_bases for base in seq_normal):
        errors.append("invalid_characters")
  return {"sequence": seq_normal,
        "valid": len(errors) == 0,
        "errors": errors,}
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """this function is to check mutations."""
    # Write your code here.

    mutations = []
    for idx, (read_base, ref_base) in enumerate(zip(sequence, reference), start=1):
        if read_base != ref_base:
            mutations.append({
                "position": idx,
                "reference_base": ref_base,
                "read_base": read_base,
            })
    return mutations
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """Replace this with a docstring describing the function."""
    # Write your code here.
    analysed_results = []

    for read in reads:
        validation = validate_read(read["sequence"], reference)
        is_valid = validation["valid"]
        clean_seq = validation["sequence"]

        # Only search for mutations if the read passed validation
        mutations = find_mutations(clean_seq, reference) if is_valid else []

        analysed_results.append({
            "read_id": read["read_id"],
            "sample": read["sample"],
            "group": read["group"],
            "sequence": clean_seq,
            "status": "valid" if is_valid else "invalid",
            "errors": validation["errors"],
            "mutations": mutations,
        })

    return analysed_results
````

### 💻 Enter your code here

````python
# Write your code here.
def _get_top_positions(position_counts: dict, n: int = 3) -> list:
    """
    Return the top `n` positions with the highest mismatch counts.
    Ties are broken by selecting the earlier sequence position (ascending order).
    """
    # Sort by count descending (-item[1]), then by position ascending (item[0])
    sorted_positions = sorted(position_counts.items(), key=lambda item: (-item[1], item[0]))
    return [pos for pos, _ in sorted_positions[:n]]


def _print_group_tables(group_name: str, group_samples: dict) -> None:
    """Print formatted summary and substitution tables for a single experimental group."""
    print(f"\n{'=' * 85}")
    print(f"GROUP SUMMARY: {group_name.upper()}")
    print(f"{'=' * 85}")

    # 1. Table of summarised results
    header = (
        f"{'Sample':<12} | {'Valid':>7} | {'Invalid':>7} | "
        f"{'Mismatch %':>11} | {'Events':>7} | {'Top Positions'}"
    )
    print(header)
    print("-" * len(header))

    for sample, stats in group_samples.items():
        pct_str = (
            f"{stats['mismatch_percentage']:.2f}%"
            if stats["mismatch_percentage"] is not None
            else "N/A (0 valid)"
        )
        top_pos_str = ", ".join(map(str, stats["top_positions"])) if stats["top_positions"] else "None"
        print(
            f"{sample:<12} | {stats['valid_reads']:>7} | {stats['invalid_reads']:>7} | "
            f"{pct_str:>11} | {stats['mismatch_events']:>7} | {top_pos_str}"
        )

    # 2. Table of substitution mutations (samples as columns, substitutions as rows)
    all_subs = sorted({
        sub for stats in group_samples.values() for sub in stats["substitution_counts"]
    })
    sample_names = list(group_samples.keys())

    print("\nSubstitution Mutations (Count & % per 100 valid reads):")
    if not all_subs:
        print("  No substitution mutations observed in this group.")
        return

    sub_header = f"{'Mutation':<10} | " + " | ".join(f"{s:^16}" for s in sample_names)
    print(sub_header)
    print("-" * len(sub_header))

    for sub in all_subs:
        row_cells = []
        for s in sample_names:
            count = group_samples[s]["substitution_counts"].get(sub, 0)
            valid_n = group_samples[s]["valid_reads"]
            if valid_n > 0:
                pct = (count / valid_n) * 100
                cell = f"{count} ({pct:.1f}%)"
            else:
                cell = f"{count} (N/A)"
            row_cells.append(f"{cell:^16}")
        print(f"{sub.replace('>', ' -> '):<10} | " + " | ".join(row_cells))


def summarise_samples(analysed_reads: list) -> dict:
    """
    Aggregate read validation and mutation metrics per sample and print group tables.

    Args:
        analysed_reads (list): Output list of dictionaries from `analyse_reads`.

    Returns:
        dict: Nested dictionary keyed by sample name containing summary metrics.
    """
    if not analysed_reads:
        return {}

    summary = {}

    for read in analysed_reads:
        sample = read["sample"]
        if sample not in summary:
            summary[sample] = {
                "group": read["group"],
                "valid_reads": 0,
                "invalid_reads": 0,
                "mismatch_reads": 0,
                "mismatch_percentage": None,
                "mismatch_events": 0,
                "substitution_counts": {},
                "position_counts": {},
                "top_positions": [],
            }

        entry = summary[sample]
        if read["status"] == "valid":
            entry["valid_reads"] += 1
            mutations = read["mutations"]
            if mutations:
                entry["mismatch_reads"] += 1
                entry["mismatch_events"] += len(mutations)

                for mut in mutations:
                    sub_key = f"{mut['reference_base']}>{mut['read_base']}"
                    entry["substitution_counts"][sub_key] = (
                        entry["substitution_counts"].get(sub_key, 0) + 1
                    )
                    pos = mut["position"]
                    entry["position_counts"][pos] = (
                        entry["position_counts"].get(pos, 0) + 1
                    )
        else:
            entry["invalid_reads"] += 1

    # Finalise percentages and top positions per sample
    for entry in summary.values():
        if entry["valid_reads"] > 0:
            entry["mismatch_percentage"] = (
                entry["mismatch_reads"] / entry["valid_reads"]
            ) * 100
        else:
            entry["mismatch_percentage"] = None

        entry["top_positions"] = _get_top_positions(entry["position_counts"], n=3)

    # Group samples preserving first-seen group order and print tables
    groups = {}
    for sample, stats in summary.items():
        grp = stats["group"]
        groups.setdefault(grp, {})[sample] = stats

    for grp_name, grp_samples in groups.items():
        _print_group_tables(grp_name, grp_samples)

    return summary
````

## Step 5 — Exploratory comparison of the top mismatch positions

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `sample_summary` is a dictionary in the format returned by the `summarise_samples` function
- `control_group` and `treated_group` are strings
- All samples have been analysed against the same reference sequence.

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `sample_summary`: output from `summarise_samples` function.
- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

Reuse helper functions from Step 4 where appropriate to reduce code complexity and repetition.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

### 📋 Return format

Return a list with one dictionary per selected position:

```python
[
    {
        "position": 42,
        "control_n": 3,
        "treated_n": 3,
        "control_mean": 0.5,
        "treated_mean": 2.0,
        "difference": 1.5,
        "statistic": ...,
        "p_value": ...
    }
]
```

Means are sample mismatch percentages; `difference` is **treated mean − control mean**, in percentage points. Use `None` for results that cannot be calculated.

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

### 💻 Enter your code here

````python
# Write your code here.
def compare_positions(
    sample_summary: dict,
    control_group: str = "control",
    treated_group: str = "treatment",
    seq_length: int = 200,
) -> list:
    """
    Identify the top 3 mismatch positions across control and treated groups and
    compare sample-level mismatch percentages using a two-sided Welch's t-test.

    Args:
        sample_summary (dict): Output dictionary from `summarise_samples`.
        control_group (str): Label for the control group.
        treated_group (str): Label for the treated group.
        seq_length (int): Length of the reference sequence (default 200).

    Returns:
        list: Dictionaries per top position with sample counts, group means,
              difference (treated - control), t-statistic, and p-value.
    """
    if not sample_summary:
        return []

    # Ensure array is long enough even if tested on shorter/longer custom sequences
    max_observed_pos = max(
        (max(s["position_counts"].keys(), default=0) for s in sample_summary.values()),
        default=0,
    )
    array_len = max(seq_length, max_observed_pos)

    # 1D array storing combined mismatch counts across control and treated groups (0-indexed)
    combined_counts_array = [0] * array_len

    control_samples = []
    treated_samples = []

    for sample_name, stats in sample_summary.items():
        grp = stats["group"]
        if grp not in (control_group, treated_group):
            continue

        if stats["valid_reads"] == 0:
            print(f"Note: Excluding sample '{sample_name}' ({grp}) due to 0 valid reads.")
            continue

        if grp == control_group:
            control_samples.append((sample_name, stats))
        elif grp == treated_group:
            treated_samples.append((sample_name, stats))

        for pos, count in stats["position_counts"].items():
            combined_counts_array[pos - 1] += count

    # Convert non-zero array entries back to a dict to reuse _get_top_positions()
    combined_pos_dict = {
        idx + 1: count for idx, count in enumerate(combined_counts_array) if count > 0
    }
    top_3_positions = _get_top_positions(combined_pos_dict, n=3)

    results = []
    print("\nStatistical Comparison of Top Mismatch Positions (Welch's two-sided t-test):")

    for pos in top_3_positions:
        ctrl_pcts = [
            (s["position_counts"].get(pos, 0) / s["valid_reads"]) * 100
            for _, s in control_samples
        ]
        treat_pcts = [
            (s["position_counts"].get(pos, 0) / s["valid_reads"]) * 100
            for _, s in treated_samples
        ]

        ctrl_n = len(ctrl_pcts)
        treat_n = len(treat_pcts)

        ctrl_mean = float(np.mean(ctrl_pcts)) if ctrl_n > 0 else None
        treat_mean = float(np.mean(treat_pcts)) if treat_n > 0 else None
        diff = (
            (treat_mean - ctrl_mean)
            if (treat_mean is not None and ctrl_mean is not None)
            else None
        )

        stat_res = calculate_statistics(ctrl_pcts, treat_pcts)

        pos_result = {
            "position": pos,
            "control_n": ctrl_n,
            "treated_n": treat_n,
            "control_mean": ctrl_mean,
            "treated_mean": treat_mean,
            "difference": diff,
            "statistic": stat_res["statistic"],
            "p_value": stat_res["p_value"],
        }
        results.append(pos_result)

        # Print one brief scientific line per position
        c_str = f"{ctrl_mean:.2f}%" if ctrl_mean is not None else "N/A"
        t_str = f"{treat_mean:.2f}%" if treat_mean is not None else "N/A"
        d_str = f"{diff:+.2f} pp" if diff is not None else "N/A"
        p_str = f"{stat_res['p_value']:.4g}" if stat_res["p_value"] is not None else "N/A"
        t_stat_str = f"{stat_res['statistic']:.3f}" if stat_res["statistic"] is not None else "N/A"

        print(
            f"Position {pos}: control mean = {c_str} (n={ctrl_n}), "
            f"treated mean = {t_str} (n={treat_n}), "
            f"difference = {d_str}, t = {t_stat_str}, p = {p_str}"
        )

    return results
````
